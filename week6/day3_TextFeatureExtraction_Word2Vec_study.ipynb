```python
# NLP related imports

from sklearn.feature_extraction.text import CountVectorizer
```

```python
# MD: https://scikit-learn.org/stable/modules/feature_extraction.html#text-feature-extraction
# https://scikit-learn.org/stable/modules/feature_extraction.html#common-vectorizer-usage
# Bag of Words study
vectorizer = CountVectorizer()
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
]
X1 = vectorizer.fit_transform(corpus)
X1
```

```python
X1.shape
```

```python
X1.toarray()
```

```python
analyze = vectorizer.build_analyzer()
analyze("This is a text document to analyze.") == (['this', 'is', 'text', 'document', 'to', 'analyze'])
```

```python
vectorizer.get_feature_names_out()
```

```python
vectorizer.vocabulary_.get('document')
```

```python
vectorizer.transform(['Something completely new.']).toarray()
```

```python
bigram_vectorizer = CountVectorizer(ngram_range=(1, 2),
                                    token_pattern=r'\b\w+\b', min_df=1)
analyze = bigram_vectorizer.build_analyzer()
analyze('Bi-grams are cool!') == (
    ['bi', 'grams', 'are', 'cool', 'bi grams', 'grams are', 'are cool'])
```

```python
X_2 = bigram_vectorizer.fit_transform(corpus).toarray()
X_2
```

```python
feature_index = bigram_vectorizer.vocabulary_.get('is this')
X_2[:, feature_index]
```

## Word2Vec

```python
# https://radimrehurek.com/gensim/models/word2vec.html
from gensim.test.utils import common_texts
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess
import numpy as np

model  = Word2Vec(sentences=common_texts, vector_size=100, window=5, min_count=1, workers=4)
model.save("word2vec.model")
```

```python
# Reoload the model from disk
model = Word2Vec.load("word2vec.model")
# Contine training with more data
model.train([["hello", "world"]], total_examples=1, epochs=1)
```

```python
vector = model.wv['computer']  # get numpy vector of a word
print(f"vector.shape: {vector.shape}, vector size: {vector.size}, vector type: {type(vector)}, vector dimension: {vector.ndim}\n vector: {vector}")
sims = model.wv.most_similar('computer', topn=10)  # get other similar words
print(f"10 most similar words to 'computer': {sims}")
```

```python
from gensim.models import KeyedVectors
# Store just words + their trained embeddings.
word_vectors = model.wv
word_vectors.save("word2vec.wordvectors")
# Load back with memory-mapping = read-only, shared across processes.
wv = KeyedVectors.load("word2vec.wordvectors", mmap='r')
vector = wv['computer']  # get numpy vector of a word (same as above)
print(f"vector.shape: {vector.shape}, vector size: {vector.size}, vector type: {type(vector)}, vector dimension: {vector.ndim}\n vector: {vector}")
```
